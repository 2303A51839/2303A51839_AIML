```python
def value_iteration(P, states, actions, gamma=0.9, theta=1e-5):
    V = {s: 0 for s in states}
    policy = {s: actions[0] for s in states}

    while True:
        delta = 0
        for s in states:
            action_values = []
            for a in actions:
                value = sum(prob * (reward + gamma * V[next_state])
                            for prob, next_state, reward in P[s][a])
                action_values.append((value, a))
            best_value, best_action = max(action_values)
            delta = max(delta, abs(V[s] - best_value))
            V[s] = best_value
            policy[s] = best_action
        if delta < theta:
            break
    return policy, V
```

```python
def policy_iteration(P, states, actions, gamma=0.9):
    policy = {s: actions[0] for s in states}
    V = {s: 0 for s in states}

    while True:
        # Policy Evaluation
        while True:
            delta = 0
            for s in states:
                a = policy[s]
                value = sum(prob * (reward + gamma * V[next_state])
                            for prob, next_state, reward in P[s][a])
                delta = max(delta, abs(V[s] - value))
                V[s] = value
            if delta < 1e-5:
                break

        # Policy Improvement
        policy_stable = True
        for s in states:
            old_action = policy[s]
            action_values = []
            for a in actions:
                value = sum(prob * (reward + gamma * V[next_state])
                            for prob, next_state, reward in P[s][a])
                action_values.append((value, a))
            _, best_action = max(action_values)
            policy[s] = best_action
            if old_action != best_action:
                policy_stable = False

        if policy_stable:
            break

    return policy, V
```

```python
states = ['A', 'B']
actions = ['left', 'right']
P = {
    'A': {
        'left': [(1.0, 'A', 0)],
        'right': [(1.0, 'B', 1)]
    },
    'B': {
        'left': [(1.0, 'A', 0)],
        'right': [(1.0, 'B', 0)]
    }
}

# Run Value Iteration
policy_vi, V_vi = value_iteration(P, states, actions)
print("Value Iteration Policy:", policy_vi)
print("Value Iteration Values:", V_vi)

# Run Policy Iteration
policy_pi, V_pi = policy_iteration(P, states, actions)
print("Policy Iteration Policy:", policy_pi)
print("Policy Iteration Values:", V_pi)
```
